### Visualize the most in-demand skills

```python
import pandas as pd
import matplotlib.pyplot as plt

skills_df = pd.read_csv("../dataset/skill_demand.csv")

top_skills = skills_df.head(10)

plt.figure(figsize=(10, 6))

plt.barh(
    top_skills["Skill"],
    top_skills["Job_Count"],
    edgecolor="black"
)

plt.xlabel("Number of Job Postings")
plt.ylabel("Technical Skills")
plt.title("Top 10 In-Demand Technical Skills")

plt.gca().invert_yaxis()

plt.tight_layout()
plt.show()
```
